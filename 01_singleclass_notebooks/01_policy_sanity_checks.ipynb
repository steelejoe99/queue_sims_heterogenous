# Basic GI/GI/N+GI Simulation

This notebook demonstrates the most basic usage of the simulator.

We construct a single-class queue with:

• general interarrival distribution  
• general service distribution  
• general patience distribution  
• a configurable scheduling policy  

The goal is simply to verify that the simulator runs correctly and to show how different policies can be applied.

```python
from gi_gi_n_gi_multiclass.core.engine import run_sim
from gi_gi_n_gi_multiclass.core.types import SimConfig
from gi_gi_n_gi_multiclass.dists.common import Exponential
from gi_gi_n_gi_multiclass.policies.fcfs import FCFS
from gi_gi_n_gi_multiclass.policies.lcfs import LCFS
from gi_gi_n_gi_multiclass.policies.tiq import TIQ
from gi_gi_n_gi_multiclass.metrics.basic import summary_table
```

## Example 1: FCFS scheduling

We begin with a simple overloaded system.

System parameters:

Servers: 10  
Arrival rate: 20  
Service rate: 1  
Patience rate: 0.5  

Since the arrival rate exceeds total service capacity, the system is overloaded and abandonment stabilizes the queue.

Customers are served using a standard **First-Come First-Served (FCFS)** discipline.

```python
cfg = SimConfig(
    n_servers=10,
    interarrival=Exponential(rate=20.0),
    service=Exponential(rate=1.0),
    patience=Exponential(rate=0.5),
    policy=FCFS(),
    seed=123,
    warmup_time=500.0,
    run_time=2000.0,
    collect_event_log=False,
)
res = run_sim(cfg)
summary_table(res)
```

## Example 2: LCFS scheduling

We now change the scheduling policy to **Last-Come First-Served (LCFS)**.

The arrival rate is reduced so the system operates closer to capacity.

Changing the policy allows us to observe how different service disciplines affect waiting times and abandonment behavior.

```python
cfg = SimConfig(
    n_servers=10,
    interarrival=Exponential(rate=8.0),
    service=Exponential(rate=1.0),
    patience=Exponential(rate=0.5),
    policy=LCFS(),
    seed=123,
    warmup_time=500.0,
    run_time=2000.0,
    collect_event_log=False,
)
res = run_sim(cfg)
summary_table(res)
```

## Example 3: TIQ scheduling

Finally we run the same system using a **Threshold Index Queue (TIQ)** policy.

TIQ divides waiting customers into regions determined by two thresholds:

$w_l$ — lower threshold  
$w_h$ — upper threshold  

Customers with waiting times above \(w_h\) are prioritized, while other customers are served according to the TIQ ordering rule.

This notebook does not attempt to choose optimal thresholds.  
The goal is simply to illustrate how the TIQ policy can be applied.

```python
cfg = SimConfig(
    n_servers=10,
    interarrival=Exponential(rate=8.0),
    service=Exponential(rate=1.0),
    patience=Exponential(rate=0.5),
    policy=TIQ(w_l=1.0, w_h=2.0),
    seed=123,
    warmup_time=500.0,
    run_time=2000.0,
    collect_event_log=False,
)
res = run_sim(cfg)
summary_table(res)
```
